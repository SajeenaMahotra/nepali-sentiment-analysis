```python
import pandas as pd
df = pd.read_csv('data/raw/scraped_reviews/scraped_reviews.csv')
print(df.shape)
print(df.columns.tolist())
print(df.head(3))
```

```python
import re
devanagari = df['review_text'].str.contains(r'[\u0900-\u097F]', regex=True)
print("Reviews with Devanagari script:", devanagari.sum())
```

```python
df = df[~df['review_text'].str.contains(r'[\u0900-\u097F]', regex=True)]
print("Reviews remaining:", len(df))
```

```python
print(df['sentiment_label'].value_counts())
```

```python
import re

def clean_text(text):
    text = text.lower()                          # lowercase everything
    text = re.sub(r'http\S+', '', text)          # remove URLs
    text = re.sub(r'[^a-z\s]', '', text)         # keep only letters and spaces
    text = re.sub(r'\s+', ' ', text).strip()     # remove extra whitespace
    return text

df['cleaned_text'] = df['review_text'].apply(clean_text)
print(df[['review_text', 'cleaned_text']].head(5))
```

```python
df = df[df['cleaned_text'].str.strip().str.len() > 2]
print("Reviews remaining:", len(df))
```

```python
from sklearn.model_selection import train_test_split

# First split: 85% train+val, 15% test
train_val, test = train_test_split(df, test_size=0.15, random_state=42, stratify=df['sentiment_label'])

# Second split: split train+val into 70% train, 15% val
train, val = train_test_split(train_val, test_size=0.176, random_state=42, stratify=train_val['sentiment_label'])

print("Train:", len(train))
print("Val:", len(val))
print("Test:", len(test))
```

```python
print("Train:\n", train['sentiment_label'].value_counts())
print("\nVal:\n", val['sentiment_label'].value_counts())
print("\nTest:\n", test['sentiment_label'].value_counts())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    min_df=2,
    sublinear_tf=True
)

X_train = vectorizer.fit_transform(train['cleaned_text'])
X_val   = vectorizer.transform(val['cleaned_text'])
X_test  = vectorizer.transform(test['cleaned_text'])

y_train = train['sentiment_label']
y_val   = val['sentiment_label']
y_test  = test['sentiment_label']

print("X_train:", X_train.shape)
print("X_val:  ", X_val.shape)
print("X_test: ", X_test.shape)
print("Vocabulary size:", len(vectorizer.vocabulary_))
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report

model = MultinomialNB()
model.fit(X_train, y_train)

y_pred = model.predict(X_val)
print(classification_report(y_val, y_pred))
```
